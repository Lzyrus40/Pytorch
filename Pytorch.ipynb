# **6. Creating a range and tensors_like**

**Range**

```
tensor_with_range = torch.arange(1,10)
tensor_with_range
```

```
tensor_with_range1 = torch.arange(1,10,2)
tensor_with_range1
```

**Tensors like**

```
sample = torch.rand(3,4)
sample
```

```
zero_like_tensor = torch.zeros_like(sample)
zero_like_tensor
```

```
ones_like_tensor = torch.ones_like(sample)
ones_like_tensor
```

**Torch eye**

```
# one digonal value
eye_tensor = torch.eye(2)
eye_tensor
```

**Torch Full**

```
full_tensor = torch.full((2,3),1)
full_tensor1 = torch.full((2,3),2)
print(full_tensor)
print(full_tensor1)
```
