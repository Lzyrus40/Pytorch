**2. Substraction**

```
tensor = torch.tensor([1,2,3])
tensor
```

```
tensor - 13
```

**3. Multiplication (element-wise)**

**4. Division**

**5. Matrix Multiplication**

```
matrix_a = torch.tensor([[1,2],[3,4]])
matrix_b = torch.tensor([[5,6],[7,8]])
matrix_a,matrix_b
```

```
matrix_a.shape, matrix_b.shape
```

```
# normal multiplication
matrix_a * matrix_b
```

```
# matrix multiplication
matrix_a @ matrix_b
```

```
torch.mm(matrix_a,matrix_b)
```

```
torch.matmul(matrix_a,matrix_b)
```

**6. One of the most common error in deep learning (shape error)**

```
tensor_A = torch.tensor([[1,2],[3,4],[5,6]],dtype=torch.float32)
tensor_B = torch.tensor([[7,8],[9,10],[11,12]],dtype=torch.float32)
tensor_A,tensor_B
```

```
tensor_A.shape,tensor_B.shape
```

```
tensor_A.T.shape, tensor_B.shape
```

```
torch.matmul(tensor_A.T,tensor_B)
```

# **10. Reshaping , Stacking , Squeezing , Unsqeeze , Permute**

**Reshape**

```
x = torch.arange(1.,9.)
x
```

```
x_reshaped = x.reshape(2,2,2)
x_reshaped
```

**View (It also change the main tensor rather then the copy of the tensor)**

```
y = torch.arange(1.,9.)
y
```

```
y_view = y.view(2,4)
y_view
```

**Torch.stack()**

```
tensor_stacked = torch.stack([tensor,tensor,tensor, tensor],dim=0)
tensor_stacked
```

```
tensor_stacked = torch.stack([tensor,tensor,tensor, tensor],dim=1)
tensor_stacked
```

**Torch.unsqueeze()**

```
x = torch.arange(1.,8.)
x
```

```
x.shape,x.ndim
```

```
x_unsqueezed = x.unsqueeze(dim=0)
x_unsqueezed.shape,x_unsqueezed.ndim
```

**Torch.squeeze()**

```
x_squeeze = x_unsqueezed.squeeze()
x_squeeze.shape,x_squeeze.ndim
```

**Permute**

```
x = torch.rand(2,3,5,5,6,4,8)
x
```

```
x.shape
```

```
# also change the original tensor
x_permute = x.permute(6,5,4,3,2,1,0)
x_permute.shape
```
